```python
import dotenv
from IPython.display import Image, display
from langgraph.types import Command

from src.apps.deep_research import build_graph,run_research
from rich import print as rprint
import uuid
from src.apps.deep_research import DeepResearchState

dotenv.load_dotenv(override=True)
```

```python

db_uri = "postgresql+asyncpg://postgres:123456@localhost:5432/deep_research_db"
graph = build_graph(db_uri)
display(Image(graph.get_graph().draw_mermaid_png()))
display(Image(graph.get_graph(xray=True).draw_mermaid_png()))
```

```python

user_query='深度调研 欧美国家酒店旅游业 在AI Agent 的落地使用'
input_state = DeepResearchState(user_query=user_query)

tid = str(uuid.uuid4())
base_config: dict = {"configurable": {"thread_id": tid}}

res = graph.invoke(input_state, config=base_config)
rprint(res)



```
